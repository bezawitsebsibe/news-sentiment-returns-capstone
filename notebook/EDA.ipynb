```python
# Step 1: Import functions
import sys
sys.path.append('../src')

from data_preparation import load_news_data, load_stock_data, explore_news_data, explore_stock_data
from cleaning import clean_news_data, align_news_to_stock, save_processed_data
```

```python
# Step 2: Load datasets
news_file = "../data/raw/raw_analyst_ratings.csv"
stock_folder = "../data/raw/yfinance_data"
processed_folder = "../data/processed"

news_df = load_news_data(news_file)
stocks = load_stock_data(stock_folder)
```

```python
# Step 3: Explore raw datasets
explore_news_data(news_df)
explore_stock_data(stocks)
```

```python
# Step 4: Clean news data
news_df = clean_news_data(news_df)
```

```python
# Step 5: Align news for a single stock (example: AAPL) and save
aligned_news = align_news_to_stock(news_df, stocks['AAPL'], 'AAPL')
save_processed_data(aligned_news, f"{processed_folder}/AAPL_news.csv")
```
